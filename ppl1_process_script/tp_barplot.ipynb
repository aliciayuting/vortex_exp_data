```python
import matplotlib.pyplot as plt
import numpy as np
```

```python

# Categories and labels
pipelines = ['PreFLMR', 'SpeechSense']
frameworks = ['Torch Serve', 'Ray Monolithic', 'Ray Microservice', 'Vortex Monolithic', 'Vortex Microservice']
colors = ['#444444',  '#5B8DB8', '#264E86',  '#E57373', '#A61B1B']
n_pipelines = len(pipelines)
n_frameworks = len(frameworks)


# Sample QPS values (placeholder data)
qps_values = [
    [21, 41.4, 86, 106, 43 ,104, 116],  # pipeline1
    [11.65,4, 13.6, 21.27, 4.5 ,13.76, 23.53]   # pipeline2
]

# Position settings
bar_width = 0.1
x = np.arange(n_pipelines)

# Create plot
fig, ax = plt.subplots(figsize=(10, 6))

# Define manual bar offsets to create spacing between logical groups
# TorchServe (no offset), Ray group (small offset), Vortex group (larger offset)
group_offsets = [-0.2, -0.05, 0.05, 0.2, 0.3]  # Adjust spacing as needed

# Plot each framework with custom spacing
for i in range(n_frameworks):
    bar_positions = x + group_offsets[i]
    ax.bar(bar_positions, [qps[i] for qps in qps_values],
           width=bar_width, color=colors[i], label=frameworks[i])

# Font size configuration
label_fontsize = 23
title_fontsize = 24
xtick_fontsize = 24
ytick_fontsize = 23
legend_fontsize = 20

# Axis settings
ax.set_xlabel('Pipeline', fontsize=label_fontsize)
ax.set_ylabel('Throughput (QPS)', fontsize=label_fontsize)
ax.set_title('Throughput Comparison by Serving Framework', fontsize=title_fontsize)
ax.set_xticks(x)
ax.set_xticklabels(pipelines, fontsize=xtick_fontsize)
ax.tick_params(axis='y', labelsize=ytick_fontsize)
ax.grid(True, axis='y', linestyle='--', alpha=0.7)

# Legend on the right with custom font sizes
ax.legend(
    loc='upper right',
    fontsize=legend_fontsize,
    frameon=True  # Optional: remove legend box for clean look
)
plt.tight_layout()
plt.savefig('throughput_comparison2.pdf', dpi=600)
plt.show()
```
